```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
search = GoogleSerperAPIWrapper()
```

```python
agent = create_agent(
    model = llm,
    tools = [search.run],
    system_prompt = "You are a helpful assistant that can answer questions using the Google search API.",
)
```

```python
question = "get me the contact details of the best restaurants in kharadi City"

search_results = agent.invoke({"messages": [{"role": "user", "content": question}]})
print(search_results)
```
